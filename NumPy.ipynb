```python
import numpy as np
```

```python
pecas_vendidas = np.array([150, 120, 90, 210, 300, 250, 180])

print(pecas_vendidas)
```

```python
faturamento = pecas_vendidas * 50

print(faturamento)
```

```python
mascara = pecas_vendidas > 200

print(mascara)
```

```python
dias_pico = pecas_vendidas[mascara]

print(dias_pico)
```

```python
print(pecas_vendidas[pecas_vendidas > 200])
```

```python
vendas = np.array([
    [200, 220, np.nan, 250],
    [150, 180, 160, 190],
    [300, 310, 290, 330]
])

print(vendas)
```

```python
total_lojas = np.sum(vendas, axis=1)

print(total_lojas)
```

```python
total_lojas = np.nansum(vendas, axis=1)

print(total_lojas)
```

```python
media_geral = np.nanmean(vendas)

print(media_geral)
```

```python
print(f"Média diária geral: {media_geral:.2f}")
```

```python
meta = 200
```

```python
resultado_meta = np.where(
    vendas >= meta,
    "Meta Atingida",
    "Abaixo"
)

print(resultado_meta)
```

```python
rng = np.random.default_rng(42)
```

```python
categorias = np.array([
    "Eletrônicos",
    "Roupas",
    "Casa"
])
```

```python
vendas_categorias = rng.choice(
    categorias,
    size=50
)

print(vendas_categorias)
```

```python
categorias_unicas, quantidades = np.unique(
    vendas_categorias,
    return_counts=True
)

print(categorias_unicas)
print(quantidades)
```

```python
for categoria, quantidade in zip(categorias_unicas, quantidades):
    print(categoria, ":", quantidade)
```

# Segunda atividade — Organização e Transformação

```python
dias = np.arange(1, 31)

print(dias)
```

```python
metas = np.linspace(20000, 30000, 5)

print(metas)
```

```python
estoque = np.array([10.5, 20.1, 30.9])

estoque_inteiro = estoque.astype(int)

print(estoque_inteiro)
```

```python
ultimos_dias = dias[-5:]

print(ultimos_dias)
```

```python
ultimos_dias_invertidos = dias[-5:][::-1]

print(ultimos_dias_invertidos)
```

```python
visitas = np.array([
    1000, 1200, 1500,
    1300, 1400, 1600,
    1700, 1800, 2000,
    2100, 2200, 2500
])
```

```python
visitas_trimestres = visitas.reshape(4, 3)

print(visitas_trimestres)
```

```python
primeiro_semestre = np.array([
    [1000, 1200, 1500],
    [1300, 1400, 1600]
])

segundo_semestre = np.array([
    [1700, 1800, 2000],
    [2100, 2200, 2500]
])
```

```python
ano = np.vstack([
    primeiro_semestre,
    segundo_semestre
])

print(ano)
```

```python
primeiro_trimestre = ano[0].copy()

print(primeiro_trimestre)
```

```python
primeiro_trimestre[0] = 9999

print(primeiro_trimestre)
print(ano)
```

```python
dados_achatados = ano.ravel()

print(dados_achatados)
```

```python
notas = np.array([85, 92, 78, 95, 88])
```

```python
ranking = np.argsort(notas)

print(ranking)
```

```python
ranking_decrescente = np.argsort(notas)[::-1]

print(ranking_decrescente)
```

```python
precos = np.array([100, 200, 300])
```

```python
print(precos.shape)
```

```python
precos_coluna = precos[:, np.newaxis]

print(precos_coluna)
print(precos_coluna.shape)
```

```python
descontos = np.array([0.10, 0.20, 0.30])
```

```python
precos_com_desconto = precos_coluna * (1 - descontos)

print(precos_com_desconto)
```
