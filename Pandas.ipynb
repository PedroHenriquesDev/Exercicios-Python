```python
import numpy as np
import pandas as pd

# Base de vendas
dados_vendas = {
    'cliente_id': [101, 102, 103, 101, 104, 102, 105, 103],
    'valor': [3500.75, 189.50, np.nan, 1200.00, 450.00, np.nan, 89.90, 780.50],
    'categoria': [
        'Eletronicos',
        'Livros',
        'Roupas',
        'Eletronicos',
        'Automotivo',
        'Livros',
        'Roupas',
        'Roupas',
    ],
    'data_hora': [
        '2024-01-15 10:23:00',
        '2024-01-18 14:05:00',
        '2024-02-05 09:12:00',
        '2024-02-20 16:40:00',
        '2024-03-02 11:00:00',
        '2024-03-15 18:30:00',
        '2024-04-10 08:20:00',
        '2024-04-22 13:45:00',
    ],
    'status': [
        'Concluído',
        'Concluído',
        'Pendente',
        'Concluído',
        'Cancelado',
        'Concluído',
        'Concluído',
        'Concluído',
    ],
    'email': [
        'maria@gmail.com',
        'joao@outlook.com',
        'ana@yahoo.com',
        'maria@gmail.com',
        'carlos@gmail.com',
        'joao@outlook.com',
        'lucas@empresa.com.br',
        'ana@yahoo.com',
    ],
}

# Base de clientes
dados_clientes = {
    'cliente_id': [101, 102, 103, 104, 105],
    'nome': [
        'Maria Silva',
        'Joao Souza',
        'Ana Oliveira',
        'Carlos Lima',
        'Lucas Mendes',
    ],
    'cidade': [
        'Sao Paulo',
        'Rio de Janeiro',
        'Belo Horizonte',
        'Curitiba',
        'Salvador',
    ],
}

df_vendas = pd.DataFrame(dados_vendas)
df_clientes = pd.DataFrame(dados_clientes)
```

```python
# Dimensões do DataFrame
print("Dimensões:", df_vendas.shape)

# Informações sobre colunas e tipos
df_vendas.info()
```

```python
print("Valores nulos:")
print(df_vendas.isnull().sum())
```

```python
# Mediana global
mediana_global = df_vendas['valor'].median()

# Mediana por categoria
mediana_categoria = df_vendas.groupby('categoria')['valor'].transform('median')

# Preenche os valores ausentes com a mediana da categoria.
# Caso a mediana da categoria também seja NaN, utiliza a mediana global.
df_vendas['valor'] = df_vendas['valor'].fillna(
    mediana_categoria.fillna(mediana_global)
)

print("\nValores após o preenchimento:")
print(df_vendas[['categoria', 'valor']])
```

```python
df_vendas['valor'] = df_vendas['valor'].fillna(
    df_vendas.groupby('categoria')['valor'].transform('median')
).fillna(mediana_global)
```

```python
vendas_filtradas = df_vendas.loc[
    (df_vendas['status'] == 'Concluído') &
    (df_vendas['valor'] > 500)
]

print(vendas_filtradas)
```

```python
vendas_filtradas = df_vendas.query(
    "status == 'Concluído' and valor > 500"
)

print(vendas_filtradas)
```

```python
df_completo = pd.merge(
    df_vendas,
    df_clientes,
    on='cliente_id',
    how='left'
)

print(df_completo)
```

```python
df_completo['media_categoria'] = (
    df_completo
    .groupby('categoria')['valor']
    .transform('mean')
)

print(
    df_completo[
        ['categoria', 'valor', 'media_categoria']
    ]
)
```

```python
gmail = df_vendas['email'].str.contains(
    '@gmail.com',
    case=False,
    na=False
)

print("Quantidade de registros com Gmail:", gmail.sum())
```

```python
df_completo['data_hora'] = pd.to_datetime(
    df_completo['data_hora']
)
```

```python
df_completo['mes'] = (
    df_completo['data_hora']
    .dt.month_name()
)

print(
    df_completo[
        ['data_hora', 'mes']
    ]
)
```

```python
meses = {
    1: 'Janeiro',
    2: 'Fevereiro',
    3: 'Março',
    4: 'Abril',
    5: 'Maio',
    6: 'Junho',
    7: 'Julho',
    8: 'Agosto',
    9: 'Setembro',
    10: 'Outubro',
    11: 'Novembro',
    12: 'Dezembro'
}

df_completo['mes'] = (
    df_completo['data_hora']
    .dt.month
    .map(meses)
)

print(df_completo[['data_hora', 'mes']])
```

```python
df_completo['dia_semana'] = (
    df_completo['data_hora']
    .dt.day_name()
)

print(df_completo[['data_hora', 'dia_semana']])
```

```python
tabela_pivot = pd.pivot_table(
    df_completo,
    values='valor',
    index='categoria',
    columns='cidade',
    aggfunc='sum',
    fill_value=0
)

print(tabela_pivot)
```
