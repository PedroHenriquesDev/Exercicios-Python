```python
import pandas as pd
import numpy as np
import folium
from folium.plugins import MarkerCluster

# Gerando dados sintéticos
np.random.seed(42)
n_imoveis = 45

dados_imoveis = {
    'id_imovel': range(1, n_imoveis + 1),
    'cidade': np.where(
        np.random.rand(n_imoveis) > 0.4,
        'Nova Iguaçu',
        'Queimados'
    ),
    'valor_venda': np.random.uniform(
        150000,
        850000,
        n_imoveis
    ).round(2),
    'tipo': np.random.choice(
        ['Casa', 'Apartamento', 'Terreno'],
        n_imoveis
    )
}

df_mapa = pd.DataFrame(dados_imoveis)

# Função para gerar latitude
def gerar_lat(cidade):
    if cidade == 'Nova Iguaçu':
        return -22.756 + np.random.uniform(-0.03, 0.03)
    return -22.716 + np.random.uniform(-0.02, 0.02)

# Função para gerar longitude
def gerar_lon(cidade):
    if cidade == 'Nova Iguaçu':
        return -43.460 + np.random.uniform(-0.03, 0.03)
    return -43.555 + np.random.uniform(-0.02, 0.02)

# Criando as coordenadas
df_mapa['latitude'] = df_mapa['cidade'].apply(gerar_lat)
df_mapa['longitude'] = df_mapa['cidade'].apply(gerar_lon)

# Visualizando os primeiros registros
df_mapa.head()
```

# Parte 1 — Inicialização e Marcadores Básicos
2. Criar o mapa centralizado na média das coordenadas

```python
lat_media = df_mapa['latitude'].mean()
lon_media = df_mapa['longitude'].mean()

print("Latitude média:", lat_media)
print("Longitude média:", lon_media)
```

```python
mapa_base = folium.Map(
    location=[lat_media, lon_media],
    zoom_start=12,
    tiles='OpenStreetMap'
)

mapa_base
```

# 3. Adicionar marcadores para os 5 primeiros imóveis

```python
for _, imovel in df_mapa.head(5).iterrows():

    popup_texto = f"""
    <b>Tipo:</b> {imovel['tipo']}<br>
    <b>Valor de venda:</b> R$ {imovel['valor_venda']:,.2f}
    """

    folium.Marker(
        location=[
            imovel['latitude'],
            imovel['longitude']
        ],
        popup=folium.Popup(
            popup_texto,
            max_width=300
        )
    ).add_to(mapa_base)

mapa_base
```

#Parte 2 — Customização Visual com CircleMarker
4. Criar um novo mapa com marcadores circulares

```python
mapa_circulos = folium.Map(
    location=[lat_media, lon_media],
    zoom_start=12,
    tiles='OpenStreetMap'
)
```

```python
for _, imovel in df_mapa.iterrows():

    # Define a cor de acordo com a cidade
    if imovel['cidade'] == 'Nova Iguaçu':
        cor = 'blue'
    else:
        cor = 'orange'

    popup_texto = f"""
    <b>ID:</b> {imovel['id_imovel']}<br>
    <b>Cidade:</b> {imovel['cidade']}<br>
    <b>Tipo:</b> {imovel['tipo']}<br>
    <b>Valor:</b> R$ {imovel['valor_venda']:,.2f}
    """

    folium.CircleMarker(
        location=[
            imovel['latitude'],
            imovel['longitude']
        ],
        radius=8,
        color=cor,
        fill=True,
        fill_color=cor,
        fill_opacity=0.7,
        tooltip='Clique para detalhes',
        popup=folium.Popup(
            popup_texto,
            max_width=300
        )
    ).add_to(mapa_circulos)

mapa_circulos
```

# Parte 3 — Agrupamento com Clustering
5. Criar o terceiro mapa e o MarkerCluster

```python
mapa_cluster = folium.Map(
    location=[lat_media, lon_media],
    zoom_start=12,
    tiles='OpenStreetMap'
)

cluster = MarkerCluster().add_to(mapa_cluster)
```

# 6. Adicionar os imóveis ao cluster com ícones personalizados

```python
cores_tipo = {
    'Casa': 'green',
    'Apartamento': 'blue',
    'Terreno': 'gray'
}
```

```python
for _, imovel in df_mapa.iterrows():

    cor_icone = cores_tipo[imovel['tipo']]

    popup_texto = f"""
    <b>ID do imóvel:</b> {imovel['id_imovel']}<br>
    <b>Cidade:</b> {imovel['cidade']}<br>
    <b>Tipo:</b> {imovel['tipo']}<br>
    <b>Valor de venda:</b> R$ {imovel['valor_venda']:,.2f}
    """

    marcador = folium.Marker(
        location=[
            imovel['latitude'],
            imovel['longitude']
        ],
        popup=folium.Popup(
            popup_texto,
            max_width=300
        ),
        tooltip='Clique para detalhes',
        icon=folium.Icon(
            color=cor_icone,
            icon='home',
            prefix='fa'
        )
    )

    marcador.add_to(cluster)

mapa_cluster
```

# 7. Salvar o mapa final em HTML

```python
arquivo = 'mapa_imoveis_baixada.html'

mapa_cluster.save(arquivo)

print(f'Mapa salvo com sucesso em: {arquivo}')
```
