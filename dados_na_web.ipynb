#Nível 1: Básico (URLs, Parâmetros e Cabeçalhos)

```python
import requests

# URL da API
url = "https://jsonplaceholder.typicode.com/posts"

# Exercício 1.2: parâmetros para filtrar os resultados
params = {
    "userId": 2
}

# Exercício 1.3: cabeçalho personalizado
headers = {
    "User-Agent": "MeuProjetoPython/1.0"
}

# Exercício 1.1 e 1.4: requisição GET
resposta = requests.get(
    url,
    params=params,
    headers=headers,
    timeout=10
)

# Código de status HTTP
print("Status HTTP:", resposta.status_code)

# URL final montada pelo requests
print("URL final:", resposta.url)

# Exibição dos resultados
print("\nResultados:")
print(resposta.json())
```

```python
import requests
import pandas as pd


def baixar_dados(url):
    try:
        resposta = requests.get(url, timeout=10)

        # Verifica se ocorreu algum erro HTTP
        resposta.raise_for_status()

        return resposta

    except requests.exceptions.HTTPError as erro:
        print(f"Erro HTTP ao acessar a URL: {erro}")

    except requests.exceptions.RequestException as erro:
        print(f"Falha na requisição: {erro}")

    return None


ceps = [
    "01001000",
    "20040002",
    "30130010"
]

resultados = []

for cep in ceps:
    url = f"https://viacep.com.br/ws/{cep}/json/"

    resposta = baixar_dados(url)

    if resposta is not None:
        dados = resposta.json()

        # Verifica se o ViaCEP informou que o CEP não existe
        if dados.get("erro"):
            print(f"CEP {cep} não encontrado.")
        else:
            resultados.append(dados)

# Cria um DataFrame com os resultados
df = pd.DataFrame(resultados)

print("\nDataFrame:")
print(df)


url_imagem = "https://picsum.photos/400/400"

resposta = baixar_dados(url_imagem)

if resposta is not None:
    with open("imagem_aleatoria.jpg", "wb") as arquivo:
        arquivo.write(resposta.content)

    print("\nImagem salva como 'imagem_aleatoria.jpg'")
```

```python
import requests
import pandas as pd
from bs4 import BeautifulSoup


url_robots = "https://books.toscrape.com/robots.txt"

resposta_robots = requests.get(url_robots, timeout=10)

print("Status do robots.txt:", resposta_robots.status_code)
print("\nConteúdo do robots.txt:")
print(resposta_robots.text)


url = "https://books.toscrape.com/"

headers = {
    "User-Agent": "MeuProjetoScraping/1.0"
}

resposta = requests.get(
    url,
    headers=headers,
    timeout=10
)

resposta.raise_for_status()

# Utiliza o analisador html.parser
soup = BeautifulSoup(resposta.text, "html.parser")

# Localiza os livros da página
livros = soup.select("article.product_pod")

dados = []

# Pega somente os 5 primeiros
for livro in livros[:5]:

    # Título
    titulo = livro.find("h3").find("a")["title"]

    # Preço
    preco = livro.select_one(".price_color").get_text(strip=True)

    dados.append({
        "titulo": titulo,
        "preco": preco
    })


# Exibe os resultados
df_livros = pd.DataFrame(dados)

print("\n5 primeiros livros:")
print(df_livros)


df_livros.to_csv(
    "livros.csv",
    index=False,
    encoding="utf-8-sig"
)

print("\nArquivo 'livros.csv' criado com sucesso!")
```

```python
import pandas as pd
import requests
from io import StringIO


# URL da Wikipedia
url = "https://en.wikipedia.org/wiki/List_of_countries_and_dependencies_by_population"

# Faz o download da página
resposta = requests.get(
    url,
    headers={"User-Agent": "MeuProjetoScraping/1.0"},
    timeout=10
)

resposta.raise_for_status()

# Converte o HTML para uma StringIO
html = StringIO(resposta.text)

# Lê as tabelas HTML diretamente com Pandas
tabelas = pd.read_html(html)

print("Quantidade de tabelas encontradas:", len(tabelas))

# Exibe as tabelas encontradas
for i, tabela in enumerate(tabelas):
    print(f"\nTabela {i}:")
    print(tabela.head())


# Seleciona a primeira tabela
df_populacao = tabelas[0]

print("\nDataFrame selecionado:")
print(df_populacao.head())

# Salva em CSV
df_populacao.to_csv(
    "populacao_paises.csv",
    index=False,
    encoding="utf-8-sig"
)

print("\nArquivo 'populacao_paises.csv' criado com sucesso!")
```
